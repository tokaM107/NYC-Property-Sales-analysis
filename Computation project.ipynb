```python
# --- Visualizing Distributions using For Loops ---

print("\n--- Visualizing Distributions ---")

# Histograms for numerical features
for col in numeric_cols_eda:
    plt.figure(figsize=(12, 6))
    sns.histplot(df[col].dropna(), kde=True, bins=50)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

# --- Count plots for categorical features (with improvements) ---
categorical_cols_eda = ['BOROUGH', 'NEIGHBORHOOD', 'BUILDING CLASS CATEGORY', 'TAX CLASS AT TIME OF SALE']
top_n = 50  # Choose the top N categories to display for 'BUILDING CLASS CATEGORY'

for col in categorical_cols_eda:
    plt.figure(figsize=(12, 7))  # Increased figure size for better readability
    if df[col].nunique() > top_n:
        top_categories = df[col].value_counts().nlargest(top_n).index
        sns.countplot(y=df[df[col].isin(top_categories)][col],
                      order=top_categories)
        plt.title(f'Count of Properties by Top {top_n} {col}')
    else:
        sns.countplot(y=df[col], order=df[col].value_counts().index)
        plt.title(f'Count of Properties by {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# --- Visualizing Relationships using For Loops ---

print("\n--- Visualizing Relationships ---")

# Scatter plots for numerical pairs (Sale Price vs. others)
for col in numeric_cols_eda[1:]:  # Exclude Sale Price vs. itself
    plt.figure(figsize=(12, 8))
    sns.scatterplot(x=col, y='SALE PRICE', data=df, alpha=0.3)  # Added data=df
    plt.title(f'Sale Price vs. {col}')
    plt.xlabel(col)
    plt.ylabel('Sale Price')
    plt.show()
    plt.show()
```

```python
# --- Bar plots for Sale Price vs. categorical features in subplots ---
categorical_cols_eda = df.select_dtypes(include='object').columns
top_n = 50  # Choose the top N categories to display for high-cardinality features

for col in categorical_cols_eda:
    plt.figure(figsize=(18, 8))  # Adjust figure size for two subplots

    # Subplot 1: Bar Plot of Average Sale Price
    plt.subplot(1, 2, 1)  # 1 row, 2 columns, first subplot
    if df[col].nunique() > top_n:
        top_categories = df[col].value_counts().nlargest(top_n).index
        avg_price = df[df[col].isin(top_categories)].groupby(col)['SALE PRICE'].mean().sort_values(ascending=False)
        sns.barplot(x=avg_price.index, y=avg_price.values)
        plt.title(f'Average Sale Price by Top {top_n} {col}')
        plt.xticks(rotation=90, ha='right', fontsize=8)
        plt.ylabel('Average Sale Price')
        plt.xlabel(col)
    else:
        avg_price = df.groupby(col)['SALE PRICE'].mean().sort_values(ascending=False)
        sns.barplot(x=avg_price.index, y=avg_price.values)
        plt.title(f'Average Sale Price by {col}')
        plt.xticks(rotation=90, ha='right')
        plt.ylabel('Average Sale Price')
        plt.xlabel(col)

    # Subplot 2: Count Plot
    plt.subplot(1, 2, 2)  # 1 row, 2 columns, second subplot
    if df[col].nunique() > top_n:
        top_categories = df[col].value_counts().nlargest(top_n).index
        sns.countplot(y=df[df[col].isin(top_categories)][col], order=top_categories)
        plt.title(f'Count of Top {top_n} {col}')
        plt.ylabel(col)
        plt.xlabel('Count')
        plt.yticks([]) # Remove y-axis ticks for count plot as labels are on the left
    else:
        sns.countplot(y=df[col], order=df[col].value_counts().index)
        plt.title(f'Count of {col}')
        plt.ylabel(col)
        plt.xlabel('Count')
        plt.yticks([]) # Remove y-axis ticks for count plot

    plt.tight_layout()
    plt.show()
```

```python
# --- Visualizing Trends ---

print("\n--- Visualizing Trends ---")

# Average Sale Price Over Time
sales_over_time = df.groupby(pd.Grouper(key='SALE DATE', freq='ME'))['SALE PRICE'].mean()
plt.figure(figsize=(12, 6))
sales_over_time.plot()
plt.title('Average Sale Price Over Time')
plt.xlabel('Sale Date')
plt.ylabel('Average Sale Price')
plt.grid(True)
plt.show()

# Number of Sales Over Time
sales_count_over_time = df.groupby(pd.Grouper(key='SALE DATE', freq='ME')).size()
plt.figure(figsize=(12, 6))
sales_count_over_time.plot()
plt.title('Number of Sales Over Time')
plt.xlabel('Sale Date')
plt.ylabel('Number of Sales')
plt.grid(True)
plt.show()
```

```python
# --- Function to detect outliers using IQR ---
def detect_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = series[(series < lower_bound) | (series > upper_bound)]
    return outliers

# --- Calculate and Visualize Outliers ---
print("\n--- Outlier Analysis ---")
for col in numeric_cols_eda:
    plt.figure(figsize=(10, 6))
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col} (Outliers as Individual Points)')
    plt.ylabel(col)
    plt.show()

    outliers = detect_outliers_iqr(df[col])
    num_outliers = len(outliers)
    total_data_points = len(df[col])
    percentage_outliers = (num_outliers / total_data_points) * 100

    print(f"\nColumn: {col}")
    print(f"  Number of outliers: {num_outliers}")
    print(f"  Percentage of outliers: {percentage_outliers:.2f}%")
    if num_outliers > 0:
        print(f"  First few outlier values:\n{outliers.head()}")
    else:
        print("  No outliers detected.")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

def preprocess_nyc_sales_for_tax_class(df):
    # Standardize column names
    df.columns = df.columns.str.strip().str.upper()

    # Drop clearly irrelevant columns
    drop_cols = ['EASE-MENT', 'APARTMENT NUMBER', 'ADDRESS']
    df = df.drop(columns=[col for col in drop_cols if col in df.columns], errors='ignore')

    # Convert numeric fields
    for col in ['SALE PRICE', 'LAND SQUARE FEET', 'GROSS SQUARE FEET', 'YEAR BUILT']:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # Remove invalid SALE PRICE entries (we're not predicting it, but it's noise)
    df = df[df['SALE PRICE'] > 1000]

    # Convert SALE DATE to datetime and extract YEAR
    if 'SALE DATE' in df.columns:
        df['SALE DATE'] = pd.to_datetime(df['SALE DATE'], errors='coerce')
        df['SALE YEAR'] = df['SALE DATE'].dt.year
        df['SALE MONTH'] = df['SALE DATE'].dt.month
        df = df.drop(columns=['SALE DATE'])

    # Drop rows with any missing values
    df = df.dropna()

    # Separate target
    target = 'TAX CLASS AT TIME OF SALE'
    y = df[target]
    X = df.drop(columns=[target])

    # Identify feature types
    cat_cols = X.select_dtypes(include='object').columns.tolist()
    num_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

    # Preprocessing pipeline
    preprocessor = ColumnTransformer([
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ])

    # Transform the data
    X_processed = preprocessor.fit_transform(X)

    return X_processed, y, preprocessor
```
